## 1. Data source, loading and column selection (Data Engineer)

- **Dataset:** Sicily Airbnb Listings Dataset
- **Publisher:** Inside Airbnb – https://insideairbnb.com/get-the-data/
- **License:** CC BY 4.0 (attribution to Inside Airbnb required)
- **Date of retrieval:** 27 September 2026
- **Snapshot date:** 1 to 2 July 2026 (from `last_scraped`)
- **Files used:** `listings.csv` (detailed listings, 90 columns) and `reviews.csv` (summary reviews, 2 columns), both from the same snapshot

The data files are **not** included in this repository because Inside Airbnb does not allow redistribution. Download `listings.csv.gz` (Detailed Listings) and `reviews.csv` once from the link above, unzip them, and place both files in the same folder as this notebook: the notebook reads `listings.csv` and `reviews.csv` from there. Set `PROJECT_ROOT`, `LISTINGS_CSV` or `REVIEWS_CSV` to point to another location.

Only the 14 columns needed for the price comparison are loaded. Host-identifying columns are never read into this notebook.

```python
import os
from pathlib import Path

import pandas as pd


def project_root(start=None):
    """Return the folder that holds listings.csv and reviews.csv."""
    start = Path(start or os.environ.get("PROJECT_ROOT", Path.cwd())).expanduser().resolve()
    if (start / "listings.csv").exists():
        return start
    if (start.parent / "listings.csv").exists():
        return start.parent              # notebook sits in notebooks/, data one level up
    return start


ROOT = project_root()                    # override with PROJECT_ROOT=... if needed
RAW_PATH = Path(os.environ.get("LISTINGS_CSV", ROOT / "listings.csv"))
REVIEWS_PATH = Path(os.environ.get("REVIEWS_CSV", ROOT / "reviews.csv"))
OUT_DIR = ROOT / "outputs" / "tables"
OUT_DIR.mkdir(parents=True, exist_ok=True)

print("ROOT:", ROOT)
print("listings.csv:", RAW_PATH, "| exists:", RAW_PATH.exists())
print("reviews.csv :", REVIEWS_PATH, "| exists:", REVIEWS_PATH.exists())

# 14 columns needed for the price comparison
selected_cols = [
    "id", "price", "neighbourhood_cleansed", "room_type", "property_type",
    "accommodates", "bedrooms", "bathrooms_text", "latitude", "longitude",
    "amenities", "number_of_reviews", "review_scores_rating", "minimum_nights",
]

# Read only these columns from the file (the other 76 columns are skipped)
df = pd.read_csv(RAW_PATH, usecols=selected_cols)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# Snapshot dates of the original file
print(pd.read_csv(RAW_PATH, usecols=["last_scraped"])["last_scraped"].unique())
```

**Result:** the working dataset has 56,873 rows and 14 columns (76 of the 90 original columns excluded). No rows were removed in this step.

**Columns excluded and why (also recorded in the cleaning log):**
- 10 host-identifying columns (names, IDs, profile URLs, descriptions, locations, pictures) were excluded for privacy, because the project does not need to identify hosts.
- 18 other host-related columns, including `host_is_superhost`, were excluded because the analysis focuses on listing prices and property characteristics, not on hosts. `host_is_superhost` is an evaluative label about an individual host, so it was removed to stay consistent with the commitment not to profile or evaluate hosts.
- The remaining excluded columns (listing URLs and descriptions, availability, extra review metrics, price-quote fields and similar) are not needed for the initial price comparison.

This step only selects columns. Missing values, price format and outliers are handled in the cleaning section below.

**Limitation of the data source: only one extra table could be merged.** Inside Airbnb publishes a fixed set of files for each city (listings, reviews, calendar, neighbourhood list and map). They are separate downloads, not a relational database, and apart from `reviews.csv` none of the other tables adds information that the price comparison needs:

- `calendar.csv` holds availability per listing per day and minimum/maximum nights. The file we downloaded has **no price column**, so it cannot add a price by date or season. It also has about 20.8 million rows, which is far more than the question needs.
- `neighbourhoods.csv` and `neighbourhoods.geojson` hold only area names and boundaries. The area name is already in the listings table (`neighbourhood_cleansed`).
- The detailed reviews file contains reviewer names and comment text. It is personal data, so it is not used (see Section 6).

As a result, the working dataset is the listings table plus a review summary. The analysis cannot use information that the source does not provide, such as the price of a listing in a given season or finer detail about its location. This limit is also listed in 4.5.

## 2. Merging the review table (Data Engineer)

A second table, `reviews.csv` from Inside Airbnb, is merged into the working dataset. It has two columns (`listing_id`, `date`) and one row per review. The detailed review file (with reviewer names, IDs and comment text) is **not** used, because the project does not need personal data of guests.

Steps: (1) summarise the reviews to one row per listing (count, first and latest review date), (2) merge into the working dataset by listing id with a left join, so no listing is lost, (3) check that both files come from the same snapshot.

```python
# Table 2: reviews (only listing_id and date, no reviewer data)
reviews = pd.read_csv(REVIEWS_PATH, parse_dates=["date"])
print("Review rows:", len(reviews))
print("Date range:", reviews["date"].min().date(), "to", reviews["date"].max().date())

# One row per listing: number of reviews, date of the first and the latest review
review_summary = (
    reviews.groupby("listing_id")
    .agg(reviews_count=("date", "size"),
         first_review_date=("date", "min"),
         last_review_date=("date", "max"))
    .reset_index()
)
print("Listings with at least one review:", len(review_summary))
```

```python
# Merge: keep all listings (left join), add the review summary by listing id
df_merged = df.merge(review_summary, left_on="id", right_on="listing_id",
                     how="left", validate="one_to_one")
df_merged = df_merged.drop(columns="listing_id")

print("Rows before merge:", len(df), "| after merge:", len(df_merged))

# A listing with no row in reviews.csv has no reviews
df_merged["reviews_count"] = df_merged["reviews_count"].fillna(0).astype(int)

# Checks: both files must come from the same snapshot
print("Review rows whose listing is not in the working dataset:",
      (~reviews["listing_id"].isin(df["id"])).sum())
print("Listings where reviews_count differs from number_of_reviews:",
      (df_merged["reviews_count"] != df_merged["number_of_reviews"]).sum())

df_merged[["id", "price", "number_of_reviews", "reviews_count",
           "first_review_date", "last_review_date"]].head(5)
```

**Merge decisions (for the cleaning log):**
- Join type: left join on `id` = `listing_id`, so all 56,873 listings are kept. The number of rows is the same before and after the merge.
- `validate="one_to_one"` makes pandas stop if either side has duplicate ids.
- A listing with no row in `reviews.csv` has no reviews, so `reviews_count` was set to 0 for these listings. `first_review_date` and `last_review_date` stay empty for them.
- `reviews.csv` has 14 review rows for 4 listings that are not in the working dataset. They are not used by the left join.
- Check: for every listing, `reviews_count` equals the original `number_of_reviews`, which confirms that both files come from the same snapshot.

## Part 1. Load data and rebuild `df_clean`

The next cell repeats the cleaning log step by step. It starts from the **merged** table (`df_merged`, 56,873 rows x 17 columns), so the three review columns survive into `df_clean`. This is the **only** cleaning step in the notebook: D3 loads `df_clean` instead of cleaning the data a second time.

The `assert` lines compare the result with the numbers written in the log, so any difference stops the notebook instead of passing silently.

```python
KEEP = ["id", "neighbourhood_cleansed", "latitude", "longitude", "property_type", "room_type", "accommodates",
        "bathrooms_text", "bedrooms", "amenities", "price", "minimum_nights", "number_of_reviews",
        "review_scores_rating"]                                           # the 14 columns kept (column selection)
raw = pd.read_csv(RAW_PATH, usecols=KEEP)
assert len(raw) == 56873 and raw["id"].is_unique                          # steps 1-2

# Start from the merged table so the review columns are kept in df_clean.
assert set(["reviews_count", "first_review_date", "last_review_date"]).issubset(df_merged.columns)
df = df_merged.copy()
assert len(df) == 56873

df["price"] = pd.to_numeric(df["price"].astype(str).str.replace(r"[\$,]", "", regex=True), errors="coerce")   # step 2
df = df[df["price"].notna()].copy()                                       # step 3
assert len(df) == 51864

def bath(t):                                                              # step 4
    if pd.isna(t): return np.nan
    m = re.search(r"\d+(\.\d+)?", t)
    return float(m.group()) if m else 0.5                                  # "Half-bath" -> 0.5
df["bathrooms"] = df["bathrooms_text"].apply(bath)
df["amenities_count"] = df["amenities"].apply(lambda s: len(json.loads(s)))   # step 5
df.loc[(df["review_scores_rating"] < 1) | (df["review_scores_rating"] > 5), "review_scores_rating"] = np.nan  # step 7
df["minimum_nights_flag"] = df["minimum_nights"] > 365                    # step 10

ppg = df["price"] / df["accommodates"]                                    # steps 12-14 use log10 of price per guest
lg = np.log10(ppg)
q1, q3 = lg.quantile([0.25, 0.75]); iqr = q3 - q1
high_fence, low_fence = 10 ** (q3 + 3 * iqr), 10 ** (q1 - 3 * iqr)
df["high_price_per_guest_flag"] = ppg > high_fence                        # step 12
df["low_price_per_guest_flag"] = ppg < low_fence                          # step 13
g = lg.groupby(df["room_type"]); a, b = g.quantile(0.25), g.quantile(0.75)
room_fence = 10 ** (b + 1.5 * (b - a))
df["price_outlier"] = ppg > df["room_type"].map(room_fence)               # step 14

df_clean = df.copy()
assert df_clean.shape == (51864, 23)                                      # 17 in + 6 derived columns
assert round(high_fence, 2) == 507.97 and round(low_fence, 2) == 2.58
assert df_clean["high_price_per_guest_flag"].sum() == 40
assert df_clean["low_price_per_guest_flag"].sum() == 14
assert df_clean["price_outlier"].sum() == 669
print("df_clean matches the cleaning log:", df_clean.shape)
print("review columns kept:", [c for c in ["reviews_count", "first_review_date", "last_review_date"] if c in df_clean.columns])
print("room-type fences (USD per guest per night):", room_fence.round(2).to_dict())
```

### Helper columns for the analysis

These are *working columns* for this step only, not part of the cleaning log.

- `price_per_guest` — main price measure.
- `bedrooms_grp` — bedrooms capped at 4 (4 means "4 or more"). `NaN` stays `NaN`.
- `peer_median_amenities` — median `amenities_count` of **peers** (same room type and same `bedrooms_grp`). Used to ask whether a high price comes with more amenities than similar listings.
- `any_flag` — listing carries at least one flag (high, low or room-type outlier).

```python
d = df_clean.copy()
d["price_per_guest"] = d["price"] / d["accommodates"]
d["bedrooms_grp"] = d["bedrooms"].clip(upper=4)
d["peer_median_amenities"] = d.groupby(["room_type", "bedrooms_grp"])["amenities_count"].transform("median")
d["amenities_vs_peers"] = d["amenities_count"] - d["peer_median_amenities"]
d["reviews_bucket"] = pd.cut(d["number_of_reviews"], [-1, 0, 9, 10**6], labels=["0", "1-9", "10+"])
d["any_flag"] = d["price_outlier"] | d["high_price_per_guest_flag"] | d["low_price_per_guest_flag"]
print(d[["price_outlier", "high_price_per_guest_flag", "low_price_per_guest_flag", "any_flag"]].sum())
print("bedrooms missing (no peer group):", int(d["bedrooms"].isna().sum()))
```

## Part 2. Filtering

A filter keeps only the rows that satisfy a condition. All filters below start from the flagged listings, because that is where the project question lives. They **overlap** (one listing can be in several), so their row counts must not be added up.

```python
SHOW = ["id", "neighbourhood_cleansed", "property_type", "room_type", "accommodates", "bedrooms",
        "amenities_count", "peer_median_amenities", "number_of_reviews", "review_scores_rating",
        "price", "price_per_guest"]
out = d[d["price_outlier"]]

filters = {
    "F1_outliers":     (out,                                                                    "price_outlier = True (above room-type fence, step 14)"),
    "F2_high_flag":    (d[d["high_price_per_guest_flag"]],                                      "far-out high price per guest (step 12)"),
    "F3_low_flag":     (d[d["low_price_per_guest_flag"]],                                       "far-out low price per guest (step 13)"),
    "F4_not_explained":(out[out["amenities_vs_peers"] <= 0],                                    "outlier AND amenities <= median of peers: price not explained by amenities"),
    "F5_credible":     (out[(out["number_of_reviews"] >= 10) & (out["review_scores_rating"] >= 4.5)], "outlier AND >=10 reviews AND rating >=4.5: likely a genuine premium listing"),
    "F6_no_reviews":   (out[out["number_of_reviews"] == 0],                                     "outlier AND zero reviews: no guest evidence for the price"),
    "F7_top20":        (out.nlargest(20, "price_per_guest"),                                    "20 highest prices per guest among outliers"),
}
filter_summary = pd.DataFrame([(k, v[1], len(v[0]), round(len(v[0]) / len(d) * 100, 2)) for k, v in filters.items()],
                              columns=["filter", "condition", "rows", "% of df_clean"])
filter_summary
```

## Part 3. Group-by and aggregation

*Group-by* splits the rows into groups (by room type, area, number of bedrooms...). *Aggregation* turns each group into a few numbers. The same `stats()` function is used everywhere so all tables have the same columns and can be compared.

Column meaning: `listings` = rows in the group; `*_ppg` = price per guest per night; `outliers` / `outlier_rate_%` = listings with `price_outlier = True`.

```python
def stats(g):
    return pd.Series({
        "listings": len(g),
        "median_price": g["price"].median(),
        "median_ppg": g["price_per_guest"].median(),
        "mean_ppg": g["price_per_guest"].mean(),
        "std_ppg": g["price_per_guest"].std(),
        "min_ppg": g["price_per_guest"].min(),
        "max_ppg": g["price_per_guest"].max(),
        "median_amenities": g["amenities_count"].median(),
        "outliers": int(g["price_outlier"].sum()),
        "outlier_rate_%": g["price_outlier"].mean() * 100,
    })
def by(df, key):
    return df.groupby(key, observed=True).apply(stats, include_groups=False).reset_index()
```

## Part 4. Sensitivity check — with and without flagged listings

The cleaning log says all flagged listings are kept and that the analysis is run with and without them. If medians and rankings barely move, the conclusions do not depend on the flagged listings.

```python
clean_only = d[~d["any_flag"]]
G7 = (by(d, "room_type")[["room_type", "listings", "median_ppg", "mean_ppg"]]
      .merge(by(clean_only, "room_type")[["room_type", "listings", "median_ppg", "mean_ppg"]],
             on="room_type", suffixes=("_all", "_without_flags")))
G7["median_change_%"] = (G7["median_ppg_without_flags"] / G7["median_ppg_all"] - 1) * 100
G7["mean_change_%"] = (G7["mean_ppg_without_flags"] / G7["mean_ppg_all"] - 1) * 100
display(G7.round(2))

# do areas keep the same ranking by median price per guest?
a_all = d.groupby("neighbourhood_cleansed")["price_per_guest"].agg(["size", "median"])
a_wo = clean_only.groupby("neighbourhood_cleansed")["price_per_guest"].median()
cmp = a_all[a_all["size"] >= 100].join(a_wo.rename("median_without_flags"), how="inner")
rho = cmp["median"].corr(cmp["median_without_flags"], method="spearman")
print(f"Spearman rank correlation of area medians, with vs without flags: {rho:.3f} ({len(cmp)} areas)")
```

## Decision question, user and success criteria

**Decision question.** For each room-rental listing, should the platform flag it for price review or leave it unflagged, based on whether its price is unusually high compared with similar listings in the same area?

**Who makes the decision.** The pricing or marketplace team of the room-rental platform.

**What they need to decide.** Which listings should be flagged for further price review because their nightly price appears unusually high compared with similar listings in the same area.

**Options.** For each listing, the platform can:
1. **Flag** the listing for price review when its price appears unusually high compared with similar listings in the same area.
2. **Do not flag** the listing when its price is within the expected range compared with similar listings.
3. **Do not make a judgement** when there is not enough comparable data to make a reliable comparison.

In this notebook the three options are the three statuses `flagged`, `not flagged` and `not enough data` (Section 3).

**Criteria used to evaluate the options.**
- The listing's nightly price.
- The typical nightly price of similar listings in the same area.
- Room type (Entire home/apt, Private room, Shared room, Hotel room).
- The availability of enough comparable listings.
- The difference between the listing's actual price and its expected price based on comparable listings.

**User and use of the result.** The main stakeholder is the pricing/marketplace team. They can use the analysis to identify listings whose prices may need further review, compare each listing with similar listings in the same area, prioritise listings for manual price review, avoid flagging listings whose price is within the expected range, and avoid making a decision when there is not enough comparable data. The flag is a screening signal: it says "look here", not "this listing is overpriced".

**Why the decision matters.** It helps the platform find listings with potentially unusual prices while taking into account differences between areas and room types. The goal is a consistent price-review process based on comparable listings, rather than comparing all listings against one overall price level.

**Metric and comparison group.** `price_per_guest = price / accommodates` (USD per guest per night). A listing is compared with listings in the same `neighbourhood_cleansed` x `room_type`; a group needs at least 20 listings, otherwise the room-type fence is used.

**Success criteria (from the D1 proposal).** The project is useful if it can:
1. Identify listings whose prices are substantially higher than comparable listings in the same neighbourhood.
2. Provide a clear and reproducible definition of "potentially overpriced".
3. Compare listings while considering neighbourhood, room type and other relevant variables available in the dataset.
4. Provide a clear measure of the price difference, such as the difference between the actual listing price and the expected price for comparable listings.
5. Avoid a strong judgement when there are too few comparable listings to support a reliable comparison.
6. Produce results that the pricing/marketplace team can understand and use to identify listings for further review.

## 2. Descriptive statistics by area and room type (Nguyên)

Before testing any flag, describe the data the flags are built on: how many listings each area x room-type group holds, and where its price per guest sits. `Q1` and `Q3` are the 25th and 75th percentiles, so half of the listings in a group fall between them.

Only groups with at least 20 listings are shown (decision 3). The table is sorted by group size, because small groups produce unstable medians.

```python
desc = df_clean.copy()
desc["price_per_guest"] = desc["price"] / desc["accommodates"]
desc["bedrooms_grp"] = desc["bedrooms"].clip(upper=4)      # 4 means "4 or more"

area_room = (
    desc.groupby(["neighbourhood_cleansed", "room_type"], observed=True)
    .agg(
        listings=("id", "size"),
        q1_ppg=("price_per_guest", lambda s: s.quantile(0.25)),
        median_ppg=("price_per_guest", "median"),
        q3_ppg=("price_per_guest", lambda s: s.quantile(0.75)),
        missing_bedrooms_pct=("bedrooms", lambda s: s.isna().mean() * 100),
    )
    .reset_index()
)
area_room = area_room[area_room["listings"] >= 20].sort_values("listings", ascending=False).reset_index(drop=True)
print(f"{len(area_room)} area x room-type groups with at least 20 listings "
      f"({int(area_room['listings'].sum()):,} of {len(desc):,} listings)")
area_room.head(15).round(2)
```

```python
# How sensitive is the group definition to the minimum group size?
sizes = desc.groupby(["neighbourhood_cleansed", "room_type"], observed=True).size()

rows = []
for k in [5, 10, 20, 50, 100]:
    kept = sizes[sizes >= k]
    rows.append({
        "min_group_size": k,
        "groups_kept": len(kept),
        "listings_kept": int(kept.sum()),
        "share_of_listings_%": round(kept.sum() / len(desc) * 100, 2),
    })
group_size_table = pd.DataFrame(rows)
group_size_table
```

**What this table shows.**
- Most listings sit in a small number of large groups: entire homes and private rooms in the main tourist areas. Those are the groups where a median is stable.
- Raising the minimum group size from 20 to 100 drops many small groups but keeps most listings (see the second table). That is why 20 is used for the flag and 100 is used only when a ranking is read.
- `missing_bedrooms_pct` explains a limitation of the flag: in private rooms and hotel rooms a large share of listings has no bedroom count, so they cannot be compared with peers on amenities and end up as `not enough data`.

**Note:** the `price_outlier` flag in step 14 uses a fence calculated separately for each room_type, resulting in 669 flagged listings. The later analysis recalculates the fence for each area × room_type group, using the room-type fence as a fallback when a group contains fewer than 20 listings. This later method identifies 798 listings above the fence.

## 3. Expected price, price gap and flag status (Tân)

**Question:** for each listing, how far is its price per guest from what is typical for comparable listings, and is it flagged as *possibly overpriced*?

This section follows the group decisions (1 to 6) and runs on `df_clean` loaded in Section 1.

| Decision | What it means here |
|---|---|
| 1. Price measure | `price_per_guest` = `price / accommodates` (USD per guest per night) |
| 2. Comparison group | `neighbourhood_cleansed` × `room_type`. Bedrooms are **not** part of the group key |
| 3. Minimum group size | A comparison group needs at least 20 listings. Smaller groups use the room-type group instead (decision 4 fallback) |
| 4. Flagged | Price per guest is unusually high within its comparison group **and** not explained by amenities |
| 5. Statistic | Median (right-skewed data); the fence is computed on log10 of price per guest |

**Status rules (checked in this order, so the three statuses never overlap):**
1. `not enough data`: `bedrooms` is missing, so the listing cannot be compared with peers of the same size on amenities.
2. `flagged`: above the group fence AND `amenities_count` is not above the median of its peers (same room type and same `bedrooms_grp`, 4 means 4 or more).
3. `not flagged`: every other listing.

A flag marks an unusual price within a comparison group. It does not prove that the listing is overpriced.

```python
MIN_GROUP = 20            # decision 3
GROUP_KEYS = ["neighbourhood_cleansed", "room_type"]   # decision 2

p = df_clean.copy()
p["price_per_guest"] = p["price"] / p["accommodates"]     # decision 1
p["log_ppg"] = np.log10(p["price_per_guest"])
assert p["price_per_guest"].notna().all() and (p["price_per_guest"] > 0).all()
print("listings:", len(p))
```

### 3.1 Expected price and fence per comparison group

For every group (area × room type) we compute the median and the fence `Q3 + 1.5 x IQR` on log10 of price per guest. Groups with fewer than 20 listings take the values of their **room-type** group (decision 4 fallback).

```python
def group_stats(g):
    q1, q3 = g["log_ppg"].quantile(0.25), g["log_ppg"].quantile(0.75)
    return pd.Series({"median_log": g["log_ppg"].median(), "fence_log": q3 + 1.5 * (q3 - q1)})

area_stats = p.groupby(GROUP_KEYS).apply(group_stats, include_groups=False)
room_stats = p.groupby("room_type").apply(group_stats, include_groups=False)

p["group_size"] = p.groupby(GROUP_KEYS)["id"].transform("size")
p = p.join(area_stats, on=GROUP_KEYS)

# Fallback: groups with fewer than 20 listings use the room-type values
small = p["group_size"] < MIN_GROUP
p["used_fallback"] = small
for c in ["median_log", "fence_log"]:
    p.loc[small, c] = p.loc[small, "room_type"].map(room_stats[c])

p["expected_ppg"] = 10 ** p["median_log"]            # expected price per guest (median of the group)
p["fence_ppg"] = 10 ** p["fence_log"]
p["price_gap"] = p["price_per_guest"] - p["expected_ppg"]          # USD per guest per night
p["price_gap_pct"] = p["price_gap"] / p["expected_ppg"] * 100
p["above_fence"] = p["log_ppg"] > p["fence_log"]      # compare in log space so listings exactly on the fence are not flagged by rounding

n_groups = p.groupby(GROUP_KEYS).ngroups
n_small = p[small].groupby(GROUP_KEYS).ngroups
print(f"{n_groups} comparison groups; {n_small} have fewer than {MIN_GROUP} listings "
      f"({int(small.sum()):,} listings use the room-type fallback)")
print("above group fence:", int(p["above_fence"].sum()))
```

### 3.2 Amenities check and status

Peers for the amenities check are listings with the same room type and the same `bedrooms_grp`. A high price is *explained by amenities* when the listing has more amenities than the peer median.

```python
p["bedrooms_grp"] = p["bedrooms"].clip(upper=4)          # 4 means "4 or more"
p["peer_median_amenities"] = p.groupby(["room_type", "bedrooms_grp"])["amenities_count"].transform("median")
p["explained_by_amenities"] = p["amenities_count"] > p["peer_median_amenities"]

no_data = p["bedrooms"].isna()                                      # status 1
flagged = ~no_data & p["above_fence"] & ~p["explained_by_amenities"]   # status 2

p["status"] = np.select([no_data, flagged], ["not enough data", "flagged"], default="not flagged")
```

### 3.3 Three-status table (feeds Chart 4)

### 3.4 Checks and notes for the report

## 4. Main statistical relationship (Tân)

**Method (decision 5):** compare groups using the **median** (with Q1 and Q3) and the **outlier rate** as the main result, and add a **simple correlation** as a supporting result. The median is used because price per guest is strongly right-skewed.

**Variables (decisions 1 to 4, from Section 3):**
- Price: `price_per_guest` (USD per guest per night).
- Comparison group: `neighbourhood_cleansed` x `room_type`. Groups need at least 20 listings to be shown, and at least 100 listings when a ranking is read (decision 3).
- `outlier_rate_%`: share of listings above their group fence (all listings in the group).
- `flagged_rate_%`: share of flagged listings among listings with a known bedroom count, because a listing with no bedroom count cannot be flagged (`not enough data`).
- Correlation: Spearman rank correlation (robust to skew and outliers) between price and amenities or size.

**Reading rule:** every result describes an association in this one snapshot. None of it shows that a variable causes a higher or lower price. This section requires `p` from Section 3.

### 4.1 Groups compared: room type

```python
def compare(g):
    known = g[g["status"] != "not enough data"]          # listings with a bedroom count
    return pd.Series({
        "listings": len(g),
        "q1_ppg": g["price_per_guest"].quantile(0.25),
        "median_ppg": g["price_per_guest"].median(),
        "q3_ppg": g["price_per_guest"].quantile(0.75),
        "above_fence": int(g["above_fence"].sum()),
        "outlier_rate_%": g["above_fence"].mean() * 100,
        "listings_with_bedrooms": len(known),
        "flagged": int((g["status"] == "flagged").sum()),
        "flagged_rate_%": (known["status"] == "flagged").mean() * 100 if len(known) else np.nan,
    })

T9_room_type = p.groupby("room_type").apply(compare, include_groups=False).reset_index()
T9_room_type["missing_bedrooms_%"] = (1 - T9_room_type["listings_with_bedrooms"] / T9_room_type["listings"]) * 100
T9_room_type.round(2)
```

### 4.2 Groups compared: area × room type

```python
T9_group = p.groupby(GROUP_KEYS).apply(compare, include_groups=False).reset_index()
T9_group = T9_group[T9_group["listings"] >= MIN_GROUP].reset_index(drop=True)     # decision 3: at least 20
big = T9_group[T9_group["listings"] >= 100]                                         # decision 3: at least 100 to read a ranking
print(len(T9_group), "groups with at least 20 listings;", len(big), "with at least 100 "
      f"(they hold {int(big['listings'].sum()):,} of {len(p):,} listings)")
print("outlier rate across the groups with 100+ listings:", big["outlier_rate_%"].describe().round(2).to_dict())

cols = ["neighbourhood_cleansed", "room_type", "listings", "median_ppg", "outlier_rate_%", "flagged", "flagged_rate_%"]
print("\nHighest median price per guest (groups with 100+ listings)")
display(big.sort_values("median_ppg", ascending=False)[cols].head(8).round(2))
print("Highest outlier rate (groups with 100+ listings)")
display(big.sort_values("outlier_rate_%", ascending=False)[cols].head(8).round(2))
print("Groups with 100+ listings and at most 5 flagged listings:", int((big["flagged"] <= 5).sum()), "of", len(big))
print("Spearman, median price vs outlier rate across these groups:",
      round(big[["median_ppg", "outlier_rate_%"]].corr(method="spearman").iloc[0, 1], 3))
```

### 4.3 Simple correlation (Spearman)

Three pairs, each for all listings and inside each room type. `rho_without_flagged` repeats the result after removing the flagged listings (the check promised in the cleaning log). With this many listings even a tiny rho is statistically significant, so the **size** of rho is what matters, not a p-value.

```python
q = p.assign(amenities_vs_peers=p["amenities_count"] - p["peer_median_amenities"])
pairs = [("price_per_guest vs amenities_count", "price_per_guest", "amenities_count"),
         ("price_per_guest vs bedrooms", "price_per_guest", "bedrooms"),
         ("price_gap_pct vs amenities_vs_peers", "price_gap_pct", "amenities_vs_peers")]

def spearman(df, a, b):
    s = df[[a, b]].dropna()
    return len(s), s[a].corr(s[b], method="spearman")

rows = []
scopes = [("All listings", q)] + [(rt, q[q["room_type"] == rt]) for rt in sorted(q["room_type"].unique())]
for scope, sub in scopes:
    for label, a, b in pairs:
        n, r = spearman(sub, a, b)
        n2, r2 = spearman(sub[sub["status"] != "flagged"], a, b)
        rows.append((scope, label, n, r, n2, r2))
T9_corr = pd.DataFrame(rows, columns=["scope", "pair", "n", "spearman_rho", "n_without_flagged", "rho_without_flagged"])
T9_corr.round(3)
```

### 4.4 Interpretation (from this run)

**Group comparison (main result)**
- **Room types differ only a little in typical price.** Median price per guest is 34.91 USD (entire homes), 39.44 (private rooms), 43.35 (hotel rooms) and 30.00 (shared rooms). Hotel rooms (178 listings) and shared rooms (60) are low-data groups, so their medians and rates move easily.
- **Outlier rates are low and similar for the two large room types:** 1.57% for entire homes and 1.36% for private rooms. Shared rooms show 11.67%, but that is 7 listings. The flagged rate (among listings with a bedroom count) is 0.94% for entire homes and 1.34% for private rooms.
- **Areas differ more than room types.** Among the 86 groups with 100 or more listings (41,993 listings), the outlier rate has a median of 1.48% but ranges from 0% to 13.84%. Median price per guest is highest for private rooms in Favignana (81.75 USD) and in the area listed as "Cefal" in the data (74.00), and for entire homes in Taormina (61.88).
- **Rankings of flagged rates are fragile.** 67 of the 86 groups have 5 or fewer flagged listings, so one listing changes a group's rate by several percentage points. Read the ranking as a place to look more closely, not as a final order.
- **A group's typical price and its outlier rate are not related across groups:** the Spearman correlation between a group's median price and its outlier rate is -0.10.

**Correlation (supporting result)**
- **Amenities and price per guest are only weakly related.** Across all listings rho is 0.05 (0.08 for entire homes, 0.02 for private rooms). More amenities go with a slightly higher price per guest, but the link is too weak to explain prices.
- **Price per guest falls slightly with more bedrooms** (rho -0.03 overall). It is stronger for private rooms (-0.36) and shared rooms (-0.38), but these rest on small or incomplete data (see limitations).
- **Relative to its group, having more amenities than peers goes with a small positive price gap:** rho is 0.10 for entire homes and about 0 for private rooms.
- **Removing flagged listings barely changes any result:** rho moves by 0.02 or less for all listings and for the two large room types, as expected because flagged listings are about 1% of the data.

### 4.5 Limitations

1. **Association only.** The correlations and group differences do not show that amenities, size or area cause price differences. Hosts set their own prices, and other factors such as location detail, season, quality or host behaviour are not in the data.
2. **A flag marks an unusual price within a group, not proof of overpricing.** It depends on the fence rule (Q3 + 1.5 × IQR on log price), the group definition and the amenities rule.
3. **Missing bedrooms limit the flagged rate.** 65.1% of private rooms and 49.4% of hotel rooms have no bedroom count, against 2.2% of entire homes. The flagged rate for private rooms uses only 3,663 of 10,495 listings, and those listings may differ from the rest.
4. **`amenities_count` measures quantity, not quality or value.** A long list of basic items counts the same as a short list of premium ones.
5. **Price per guest assumes the price is shared evenly by `accommodates`.** This is a simplification, and it probably contributes to the negative link between bedrooms and price per guest.
6. **Small groups.** Groups under 20 listings use the room-type fallback and are not shown here. Groups with 20 to 99 listings are saved in the table, but their rates are not ranked.
7. **One snapshot and listed prices.** The data are listed prices on 1 to 2 July 2026, not prices guests paid, so seasonality and discounts are not visible.
8. **Spearman measures only monotonic association** and one number cannot show differences between areas inside the same room type.
9. **Only one extra table could be merged.** The other tables from the data source (calendar, neighbourhoods, detailed reviews) add no variable the price comparison needs, so the analysis cannot use season-specific prices or finer location detail.

### 4.6 What we recommend

1. **Use the flag as a screening list, not as a verdict.** A flagged listing is unusual within its area and room type after allowing for size and amenities. It is a place to look, and the report should say so in the same sentence.
2. **Ignore rankings built on fewer than about five flagged listings.** Most groups have very few flagged listings, so one listing can move a group's rate by several percentage points.
3. **Trust the flags with guest evidence most.** A listing that is far above its group fence, has at least 10 reviews and a rating of 4.5 or more is the most credible premium case; a listing with no reviews is the weakest case.
4. **Do not use these numbers to judge hosts.** The data has no host identifiers in the analysis, and the question is about prices, not about people.

These recommendations are consistent with the success criteria in the D3 decision block: like-for-like comparison, reproducible numbers, associations only, no personal data, and a stated limit.

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

# Note: put the csv file in the same folder as this notebook, or change the path below to point to it.
FILE = "outputs/tables/F8_status_counts.csv"

# 1. Read the table (3 rows, one per status)
df = pd.read_csv(FILE)
print(df)

# 2. Sanity check: the three statuses add up to 51,864 listings
total = int(df["listings"].sum())
assert total == 51864, "Total is not 51,864, please check the file"

# 3. Display names and colours (flagged stands out, the others are neutral)
names = {"flagged": "Flagged", "not flagged": "Not flagged", "not enough data": "Not enough data"}
colors = {"flagged": "#e08e2b", "not flagged": "#3b6ea5", "not enough data": "#9a9a9a"}

# 4. Draw
fig, ax = plt.subplots(figsize=(8, 5.5))
bars = ax.bar([names[s] for s in df["status"]], df["listings"],
              color=[colors[s] for s in df["status"]], width=0.6)

# Write the count and the percentage on every bar, so even the small bar can be read
for b, (_, r) in zip(bars, df.iterrows()):
    ax.text(b.get_x() + b.get_width() / 2, b.get_height() + 600,
            f"{int(r['listings']):,}\n({r['share_%']:.2f}%)", ha="center", va="bottom", fontsize=10)

# 5. Y axis: linear scale, starting at 0 (no truncated axis, no log scale on bars)
ax.set_ylim(0, df["listings"].max() * 1.15)
ax.yaxis.set_major_formatter(FuncFormatter(lambda v, _: f"{v:,.0f}"))

ax.set_title(f"Final status of listings ({total:,} listings)")
ax.set_xlabel("Status")
ax.set_ylabel("Number of listings")

plt.tight_layout()
plt.savefig(IMG_DIR / "chart4_listing_status.png", dpi=200)
plt.show()
```
